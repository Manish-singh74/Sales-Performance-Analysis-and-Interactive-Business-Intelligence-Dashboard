```python
region_sales = data.groupby("Region")["Sales"].sum()


plt.figure(figsize=(10,4))
region_sales.plot(kind="bar", color="skyblue", edgecolor="black")

def crore_format(x, pos):
    return f"{int(x/10000000)} Cr"   
plt.gca().yaxis.set_major_formatter(FuncFormatter(crore_format))

plt.title("Region-wise Total Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
region_Profit = data.groupby("Region")["Profit"].sum()


plt.figure(figsize=(10,4))
region_Profit.plot(kind="bar", color="crimson", edgecolor="black")


def crore_format(x, pos):
    return f"{int(x/10000000)} Cr"   
plt.gca().yaxis.set_major_formatter(FuncFormatter(crore_format))

plt.title("Region-wise Total Profit")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```
